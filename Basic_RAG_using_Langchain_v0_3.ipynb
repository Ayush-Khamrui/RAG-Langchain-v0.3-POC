# Basic RAG Pipeline (Langchain v0.3)
This will teach you to create your very first LLM based chatbot using own Data using RAG (Retrieval Augmented Generation).

### Step 1: Import Libraries
First import all the necessary libraries from langchain
1. From [Document Loader](https://python.lang.chat/docs/integrations/document_loaders/) you will find the various Document Loader. Here we have used PyPDFLoader to load PDF.
2. From [Embedding models](https://python.lang.chat/docs/integrations/text_embedding/) you will find all the embedding models. Here we will use hugging face embedding model.



---


### Step 2: Load Data
Here we are loading the PDF data from which we want to retrieve our data.

```python
from langchain_community.document_loaders import PyPDFLoader
# initializing the loader and setting the path of the content
loader = PyPDFLoader(
    "/content/Ayush Khamrui Backend.pdf",
)
```

load the data from the file using .load()

```python
docs = loader.load()
docs[0]
```



---


### Step 3: Split the data (Chunking)
Some common types of Chunking are:
1.  Fixed Length Chunking
2.  Sentence Based Chunking
3.  Paragrah Based Chunking
4.  Sliding Window Chunking
5.  Semantic Chunking
6.  Recursive Chunking
7.  Context Enriched Chunking
8.  Modality Specific Chunking
9.  Agentic Chunking
10. Subdocument Chunking
11. Hybrid Chunking

Like you will find all the text splitting strategies in [Text Splitter](https://python.langchain.com/api_reference/text_splitters/)

```python
# Here we are utilizing recursive chunking
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
documents = text_splitter.split_documents(docs)
```



---
### Step 4: Initialize LLM model and Embedding model



```python
# Intialize the Embedding model
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-mpnet-base-v2")
```

```python
from langchain_huggingface import HuggingFaceEndpoint, ChatHuggingFace
from huggingface_hub import login

login(token="XXXXXXXXXXXXXXXXXXXXXXXXXXXXXXXXXXX")

llm = HuggingFaceEndpoint(
    repo_id="microsoft/Phi-3-mini-4k-instruct",
    task="text-generation",
    max_new_tokens=512,
    do_sample=False,
    repetition_penalty=1.03,
)

chat_model = ChatHuggingFace(llm=llm, verbose=True)
```



---


### Step 5: Create Embeddings and store it into vector store

```python
# Create Vector Store
from langchain_chroma import Chroma

vector_store = Chroma(
    collection_name="basic_rag",
    embedding_function=embeddings,
    persist_directory="./chroma_langchain_db",  # Where to save data locally, remove if not necessary
)
```

```python
# Create Embeddings and store them
vector_store.add_documents(documents=documents)
```



---

### Step 6: Retrieve the documents
Major Retriever include:
1. Sparse Vector Retriever
2. Dense Vector Retriever
3. Hybrid Retriever
4. Semantic Search Retriever
5. Specialized Retriever
6. Blended Retriever

Like here we are using Maximal Marginal Relevance (mmr) which is typically associated with Dense Vector Retriever.

For further details on Advanced Retrieval Techniques based on usecase
see [Retrievers](https://python.langchain.com/docs/integrations/retrievers/)


```python
retriever = vector_store.as_retriever(
    search_type="mmr", search_kwargs={"k": 1, "fetch_k": 5},llm=llm
)
```



---

### Step 6: Respose Generation using Chain

Earlier RetrievalQA was used but now as per the latest documentation it is depricated. So instead use create_retrieval_chain

```python
from langchain.chains import create_retrieval_chain
from langchain.chains.combine_documents import create_stuff_documents_chain
from langchain import hub

system_prompt = hub.pull("langchain-ai/retrieval-qa-chat")
combine_docs_chain = create_stuff_documents_chain(llm, system_prompt)
rag_chain = create_retrieval_chain(retriever, combine_docs_chain)

result = rag_chain.invoke({"input": "Who is Ayush"})
print(result["answer"])
```
